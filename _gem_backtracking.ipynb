# 💎 Magical Gem Backtracking Search
This notebook demonstrates how **Backtracking** systematically generates all possible subsets (the Power Set) of an array of items.

### 🌳 Backtracking Decision Tree
Each level of the tree represents a choice for a specific gem: either **Include (+)** it or **Exclude (-)** it.

```text
                      [ ] (Empty Chest)
                     /   \
          + Ruby    /     \  - Ruby
                   /       \
             [Ruby]         [ ]
             /    \         /    \
     +Em    /      \ -Em   /      \ +Em
           /        \     /        \
     [Ruby,Em]    [Ruby] [Em]       [ ]
       /   \       /  \   /  \     /   \
  +Sap/     \-Sap /    \ /    \   /     \-Sap
     /       \   /     /       \ /       \
 [R,E,S]   [R,E] [R,S] [R]   [E,S] [E]   [S]   [ ]
```


````python
def generate_gem_combinations(gems):
    """
    Generates all possible combinations (Power Set) of a given list of gems
    using recursive backtracking.
    """
    result = []
    
    def backtrack(index, current_combination):
        # Base Case: If we've made decisions for all gems, save this path
        if index == len(gems):
            result.append(list(current_combination))
            return
        
        # Decision 1: Include the current gem
        current_combination.append(gems[index])
        backtrack(index + 1, current_combination)
        
        # Backtrack Step: Undo our choice to clear the state for the next branch
        current_combination.pop()
        
        # Decision 2: Exclude the current gem
        backtrack(index + 1, current_combination)

    # Begin tracking from index 0 with an empty container
    backtrack(0, [])
    return result

# Run an example directly in the notebook
gems_list = ["💎 Ruby", "🟢 Emerald", "🔵 Sapphire"]
all_combinations = generate_gem_combinations(gems_list)

print(f"✨ Total Combinations: {len(all_combinations)}")
print("-" * 35)
for idx, combo in enumerate(all_combinations, 1):
    print(f"Combination {idx}: {combo}")
````

````python
import unittest

class TestGemCombinations(unittest.TestCase):
    
    def test_empty_gems(self):
        # An empty chest must yield exactly 1 combination: the empty set [[]]
        self.assertEqual(generate_gem_combinations([]), [[]])
        
    def test_single_gem(self):
        # 1 item yields 2^1 = 2 combinations
        self.assertEqual(generate_gem_combinations(["💎"]), [["💎"], []])
        
    def test_combination_count(self):
        # 3 items yield 2^3 = 8 combinations
        gems = ["A", "B", "C"]
        result = generate_gem_combinations(gems)
        self.assertEqual(len(result), 8)

# Run the test suite directly inside the Jupyter Notebook cell
print("--- Running Automated Tests ---")
unittest.main(argv=[''], verbosity=2, exit=False)
````
